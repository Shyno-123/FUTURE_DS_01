```python
subcategory_sales = (
    df.groupby("Sub-Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

subcategory_sales
```

```python
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)
subcategory_profit
```

```python
subcategory_margin=(df.groupby("Sub-Category")["Profit"].sum()
                    /df.groupby("Sub-Category")["Sales"].sum()*100).sort_values(ascending=False)
subcategory_margin
```

```python
region_sales=(df.groupby("Region")["Sales"].sum()
              .sort_values(ascending=False))
region_sales
```

```python
region_profit=(df.groupby("Region")["Profit"].sum()
               .sort_values(ascending=False))
region_profit
```

```python
region_margin=(df.groupby("Region")["Profit"].sum()
               /df.groupby("Region")["Sales"].sum()
               *100).sort_values(ascending=False)
region_margin
```
